```python
import math
import numpy as np

def gaussian_weight(x, mean, std_dev):
    return math.exp(-(x - mean)**2 / (2 * std_dev**2))

def gaussian_weighted_average(lst):
    mean = np.mean(lst)
    std_dev = np.std(lst)
    weighted_sum = 0
    weight_sum = 0

    for i, value in enumerate(lst):
        weight = math.exp(-(i - mean)**2 / (2 * std_dev**2)) # calculate the gaussian weight
        weighted_sum += value * weight
        weight_sum += weight

    if weight_sum == 0:
        raise ValueError("Cannot calculate weighted average with zero total weight")

    return weighted_sum / weight_sum

# Example usage:
data_list = [1,2,3,4,5]
mean_value = np.mean(data_list)  # Using the mean as the middle index for simplicity
std_deviation = np.std(data_list)

weighted_avg = gaussian_weighted_average(data_list)
print("Gaussian Weighted Average:", weighted_avg)
```
