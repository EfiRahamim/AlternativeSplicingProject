```python
import pandas as pd

# Assuming 'data' is your pandas DataFrame
data = pd.read_csv("/private10/Projects/Efi/AML/SplicingAnalysis_March2024/SplicingEvents/_forNEanalysis/NovelStrongBindingEpitopes_noDups.csv")
df_rank = data.melt(id_vars=['Group','Splicing Event','Peptide','ID'], 
                value_vars= data.columns[data.columns.str.endswith('_Rank')],
                var_name='HLA_Rank', 
                value_name='Rank').dropna(subset=['Rank'])
df_nM = data.melt(id_vars=['Group','Splicing Event','Peptide','ID'], 
                value_vars= data.columns[data.columns.str.endswith('_nM')],
                var_name='HLA_nM', 
                value_name='nM').dropna(subset=['nM'])

df_rank['HLA'] = df_rank['HLA_Rank'].str.split('_').str[0]
df_nM['HLA'] = df_nM['HLA_nM'].str.split('_').str[0] 

df_rank.drop(columns=['HLA_Rank'], inplace=True)
df_nM.drop(columns=['HLA_nM'], inplace=True)

# Rearrange columns in each melted DataFrame
df_rank = df_rank[['Group','Splicing Event','Peptide','ID', 'HLA', 'Rank']]
df_nM = df_nM[['Group','Splicing Event','Peptide','ID', 'HLA', 'nM']]

df_merged = pd.merge(df_rank, df_nM, on=['Group','Splicing Event','Peptide','ID', 'HLA'], how='inner')

df_merged.to_csv('/private10/Projects/Efi/AML/SplicingAnalysis_March2024/SplicingEvents/_forNEanalysis/merge_test.csv', index=False)
```
